```python
#### outlier capping
def outiler_capping(df,column):
    Q1= df[column].quantile(0.25)
    Q3= df[column].quantile(0.75)
    IQR=Q3-Q1
    Lower_Extreme= Q1-1.5*IQR
    Upper_Extreme= Q3+1.5*IQR
    df[column]= df[column].apply(lambda x:Lower_Extreme if x<Lower_Extreme else Upper_Extreme if x>Upper_Extreme else x)
for col in df1.select_dtypes(['int','float']).columns:
    outiler_capping(df1,col)
```

```python
### Data Visualization
```

```python
# Histogram
for i in ['LB','AC','FM','UC','DL','DS','DP','ASTV','MSTV','ALTV','MLTV','Width','Tendency']:	
    sns.histplot(x = df[i],bins=30,edgecolor='white',color = 'grey',kde =True)
    plt.title(f'Distribution of {i}')
    plt.xlabel(i)
    plt.ylabel('counts')
    plt.show()
```

```python
for i in ['LB','AC','FM','UC','DL','DS','DP','ASTV','MSTV','ALTV','MLTV','Width','Tendency']:	
    sns.boxplot(x=df[i],color = 'skyblue')
    plt.title(f'Boxplot of {i}')
    plt.xlabel(i)
    plt.ylabel('counts')
    plt.show()
```

```python
### Bar Chart 
sns.countplot(x='NSP',data = df)
plt.show()
```

```python
### Pie Chart 

plt.figure(figsize=(5,5))
df['NSP'].value_counts().plot.pie(autopct='%1.1f%%',startangle=90,cmap="coolwarm",wedgeprops={'edgecolor': 'white'})
plt.title("Pie Chart of NSP (Target Variable)")
plt.show()
```

```python
### Scatter Plot
sns.scatterplot(x='LB', y='AC', data=df)
plt.title("LB vs AC")
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(7,5))
sns.heatmap(df.corr(),annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```
